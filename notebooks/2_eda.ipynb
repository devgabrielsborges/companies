```python
import pandas as pd

companies = pd.read_csv('../data/raw/companies.csv')
```

```python
import plotly.express as px

order = companies["company_size"].value_counts().index.tolist()


px.histogram(companies, x="company_size", color="company_size", category_orders={"company_size": order})
```

```python
px.violin(companies, x="capital_stock", y="company_size", color="company_size")
```
